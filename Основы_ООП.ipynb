```python
# 1
class Student:
    def __init__(self,name,course,average_grade):
        self.name=name
        self.course=course
        self.average_grade=average_grade
    def is_excellent(self):
        return self.average_grade>=4.5
    def show_info(self):
        print(f'Студент: {self.name}{'\n'}Курс: {self.course}{'\n'}Средний балл: {self.average_grade} {self.is_excellent()} ')
h=Student('Иван',2,4.5)
h.show_info()
h=Student('Марин',4,3.0)
h.show_info()
h=Student('Апелкн',1,5.0)
h.show_info()
```

```python
#2
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
    def check(self,amount):
        return amount<0
    def deposit(self,amount):
        if self.check(amount):
            raise RuntimeError('Неверное число')
        self.balance+=amount
    def withdraw(self,amount):
        if amount>self.balance and self.check(amount):
          raise RuntimeError('Недостаточно средств')
        self.balance-=amount
    def show_balance(self):
        print(f'Текущий баланс: {self.balance}')
account = BankAccount("Анна", 1000)
account.deposit(-500)
account.withdraw(300)
account.show_balance()
```

```python
#3
class Product:
    def __init__(self,name,price,amount):
        self.name=name
        self.price=price
        self.amount=amount
    def check(self,count):
        return count<0
    def restock(self,count):
        if self.check(count):
            raise RuntimeError('Неверное число')
        self.amount+=count
    def sell(self,count):
        if count>self.amount or self.check(count):
            raise RuntimeError('Недостаточно товара на складе')
        self.amount-=count
    def show_info(self):
        print(f'Товар: {self.name}{'\n'}Цена: {self.price}{'\n'}Количество: {self.amount}')
product=Product('Клавиатура',5000,10)
product.sell(3)
product.restock(2)
product.show_info()
```

```python
#5
class Temperature:
    @staticmethod
    def celsius_to_farenheit(value):
        return value*(9/5)+32
    @staticmethod
    def farenheit_to_celsius(value):
        return (value-32)*(5/9)
print(Temperature.celsius_to_farenheit(20))
```

```python
#6
class Book:
    def __init__(self,name,year,author):
        self.name=name
        self.year=year
        self.author=author
    def __str__(self):
        return f'{self.name} - {self.author}, {self.year}'
book1=Book("Мастер и Маргарита","М. Булгаков",1967)
book2=Book("Война и мир","Л. Толстой",1869)
book3=Book("Облако в штанах","В. Маяковский",1915)
print(book1,book2,book3,sep='\n')
```

```python
#7
class Rectangle:
    def __init__(self,width,height):
        self.width=width
        self.height=height
    def area(self):
        return self.height*self.width
    def __eq__(self,other):
        if not isinstance(other,Rectangle):
            return NotImplemented
        return self.area()==other.area()
    def __str__(self):
        return f'{self.area()}'
r1 = Rectangle(10, 5)
r2 = Rectangle(25, 2)
print(r1==r2)
```

```python
#8
class Character:
    def __init__(self,name,health,power,coins):
        self.name=name
        self.health=health
        self.power=power
        self.coins=coins
    def attack(self):
        return self.power
    def heal(self,value):
        self.health+=value
    def earn_money(self,value):
        self.coins+=value
    def spend_money(self,value):
        if value>self.coins:
            return NotImplemented
        self.coins-=value
    def show_info(self):
        print(f'Персонаж: {self.name}{'\n'}Здоровье: {self.health}{'\n'}Атака: {self.power}{'\n'}Монеты: {self.coins}')
hero=Character('Alex',100,15,50)
hero.show_info()
hero.heal(20)
hero.earn_money(100)
hero.spend_money(30)
hero.show_info()
```
